# Large Language Models - LLMs

I'm updating this notebook entry because it is a notebook entry and not a published report or article. Things have changed a bit in tech, but most of this content is still true.

Since the ChatGPT revolution of 2022 LLMs have become somewhat synonymous with the term "AI". A lot of the concepts and code that power these tools have been developed over the past few decades, expanding on the many innovations in computer science that followed the launch of the internet. 

It's important to remember that AI is **developed by people**. Better public understanding and more meaningful intention on the utilization of these tools is essential for future generations and businesses. A bit of healthy skepticism should always be applied. This notebook is intended to be conceptual and thought-provoking—hence open source.

The term "Large Language Model" started appearing after the "Transformer" architectures presented by [Google Research in 2017](https://research.google/blog/transformer-a-novel-neural-network-architecture-for-language-understanding/). LLMs are compromised of various components that require different [machine-learning](https://mitsloan.mit.edu/ideas-made-to-matter/machine-learning-explained) "roles", starting with a transformer:

| Component                | Role in Machine Learning                        |
| ------------------------ | ----------------------------------------------- |
| Transformer Architecture | Encodes and processes text efficiently          |
| Self-Attention           | Captures word relationships across sequences    |
| Pretraining Objective    | Predict next/missing tokens (self-supervised)   |
| Optimization             | Learns by minimizing loss via gradient descent  |
| Data                     | Provides diverse examples for language learning |
| Parameters               | Store learned patterns and associations         |
| Inference                | Applies learned patterns to generate text       |
| Fine-tuning / RLHF       | Aligns model to tasks or human values           |

## API Example

This sample will showcase the [Speech-to-Text](https://www.ibm.com/products/speech-to-text) service. We are going to explore [training methods](https://youtu.be/AEXS8TBd6ug?si=S4WMst7ku48RFhh7). 

```python
import ibm_watson
from ibm_cloud_sdk_core.authenticators import IAMAuthenticator
print(f'IBM Watson library: {ibm_watson.__version__}')
```

This block takes a mp4 (video) file and extracts the audio into a .wav file using [ffmpeg](https://ffmpeg.org/).

```python
# call this cell below to convert the mp4 file to audio, storing it in the data folder
command = 'ffmpeg -i data/herb.mp4 -ab 160k -ar 44100 -vn data/herb.wav'
subprocess.call(command, shell=True)
```

Connect to the IBM speech-to-text service using the API Key and URL managed from the Watson Studio console.

```python
authenticator = IAMAuthenticator(os.environ.get('IBM_STT_API_KEY'))
stt = ibm_watson.SpeechToTextV1(authenticator=authenticator)
stt.set_service_url(os.environ.get('IBM_STT_URL'))
```

`speaker_labels=True` asks Watson to split the audio into anonymous speakers (`Speaker 0`, `Speaker 1`, …). It will not name Herb Brooks or Mike Eruzione. That feature is tuned for turn-taking conversations (a phone call), not overlapping shouts, a whistle, and a film score under the dialogue.

The multimedia model above provides the best starting option for this 44.1 kHz movie audio in the `.wav`.

```python
with open('./data/herb.wav', 'rb') as f:
    res = stt.recognize(
        audio=f,
        content_type='audio/wav',
        model='en-US_Multimedia',
        speaker_labels=True,  # anonymous Speaker 0/1/... — would need to map character names
    ).get_result()
```

`speaker_labels` arrive in a nested array; timestamps (forced on by that flag) let us group words into turns

```python
words = [
    (start, word)
    for result in res.get('results', [])
    for word, start, _end in result['alternatives'][0].get('timestamps', [])
]
labels = res.get('speaker_labels', [])
```

Look up which speaker is talking during a specific time slot

```python
def speaker_for(start):
    for label in labels:
        if label['from'] <= start < label['to'] or abs(label['from'] - start) < 0.05:
            return label['speaker']
    return '?'
```

Build new array to print formatted speaker turns

```python
turns = []
for start, word in words:
    spk = speaker_for(start)
    if not turns or turns[-1][0] != spk:
        turns.append([spk, word])
    else:
        turns[-1][1] += f' {word}'
```

```python
print('\n'.join(f'Speaker {spk}: {text}' for spk, text in turns) or '(no transcript returned)')
```

## Observations

Computers can struggle with complex tasks. All the pre-trained examples in this notebook are great at processing text when it is given to them in a clear, readable format (e.g. sending typed text directly to the model, even phone audio is a relatively easy task).

Transcribing text from different formats like a movie? More difficult. Data loss occurs because Speech-to-Text is trying to turn messy sound into words, and when it is unsure it guesses. This is where [model training](https://www.ibm.com/think/topics/training-data) and evaluation matter a lot.

I intentionally picked a video with the following characteristics to highlight difficulties the Watson algorithm is wrestling with: 
- Music playing in the background
- Multiple speakers with different accents
- Muffled and difficult-to-decipher moments - The response completely butchers Mike Eruzione's epic *"I play for the United States of America!"* :(

These tasks require **human** time and energy ∴ money to make useful.
