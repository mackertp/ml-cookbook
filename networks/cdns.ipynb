# Content Delivery Network

One of the quickest ways to improve performance of a web-based applicaiton is by implementing a Content Delivery Network (CDN). This can be viewed as "intermediary" hardware that provides a performance boost to existing software systems.

It provides dedicated static storage that stores copies (caches) your site's files (including images, videos, html, and css). When someone visits your deployed website, the CDN will serve the files pre-loaded.

Most cloud service providers include a CDN option, a few are shared below: 

- [Cloudflare](https://www.cloudflare.com/learning/cdn/what-is-a-cdn/)
- [CloudFront](https://aws.amazon.com/cloudfront/getting-started/S3/)
- [Google Cloud CDN](https://cloud.google.com/cdn)

CDNs are a great practical example of how [distributed computing](https://en.wikipedia.org/wiki/Distributed_computing) and [computer networking](https://en.wikipedia.org/wiki/Computer_network) provide performance improvements.
