Once partner acess is initiatilized, run the Flask application from this directory. You will need two terminals. In the first run:

```console
uname@os:~ poetry env activate 
uname@os:~ python app.py
```

Then, in the second terminal:

```console
uname@os:~ ngrok http 8080
```

Go to your unique development web address provided by ngrok and login with Tesla. After you complete the login step from the Flask app, a token data will be stored on your local .env file. This is used to run requests from notebooks and persist access.

You can create your own notebooks or expand the Flask application from here ~ just keep that tokens file secured.
